## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
!git clone https://github.com/joj48/Flyrank.ai-SEO-Project.git
import os
os.chdir('Flyrank.ai-SEO-Project') # Navigate into the cloned repository

import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

n_rows = len(df)
n_clients = df["client_id"].nunique()
n_visible_500 = (df["impressions_90d"] >= 500).sum()
pct_visible_500 = 100 * n_visible_500 / n_rows
n_down = (df["trend_direction"].astype(str).str.lower() == "down").sum()
pct_down = 100 * n_down / n_rows

print(f"Rows (pages): {n_rows:,}")
print(f"Unique clients: {n_clients:,}")
print(f"Pages with impressions_90d >= 500: {n_visible_500:,} ({pct_visible_500:.1f}%)")
print(f"Pages with trend_direction == 'down': {n_down:,} ({pct_down:.1f}%)")
n_visible_and_down = ((df["impressions_90d"] >= 500) & (df["trend_direction"].str.lower() == "down")).sum()
print(f"Visible + down pages: {n_visible_and_down:,}")
```
